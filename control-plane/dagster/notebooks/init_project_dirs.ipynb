```python
from pathlib import Path

def init_project_dirs(base_dir: str):
    """
    Erstellt die Projektverzeichnisstruktur unter base_dir.
    """
    base_path = Path(base_dir)
    
    # Definition der Verzeichnisstruktur
    # Key: Hauptverzeichnis, Value: Liste der Subverzeichnisse
    dir_structure = {
        # experiments, pipelines, builds, 
        "logs": ["build", "run"],
        "data": ["raw", "processed"],
        "configs": [], #["pipeline", "dag", "component"],
        "artifacts": [], #["models", "reports", "metrics"],
        "runs": []  # Run-spezifische Unterordner (runXXX: config, results, components, metadata, logs) werden dynamisch erzeugt
    }
    
    for main_dir, sub_dirs in dir_structure.items():
        main_path = base_path / main_dir
        main_path.mkdir(parents=True, exist_ok=True)
        print(main_path)
        for sub in sub_dirs:
            print(main_path / sub)
            (main_path / sub).mkdir(exist_ok=True)
    
    return base_path, dir_structure

base_dir = "projects/sentiment"
base_path, dir_structure = init_project_dirs(base_dir)
```
